```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
# ==========================================
# Q1. (Data Preparation)
# ==========================================
# Load the cleaned emotions dataset (or load train.txt and apply cleaning)
df = pd.read_csv('cleaned_emotions.csv')

# Handle any missing values if present in cleaned_text
df['cleaned_text'] = df['cleaned_text'].fillna('')

# Separate features (X) and target (y)
X = df['cleaned_text']
y = df['encoded_emotions']

# Perform train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Print the shapes
print("--- Q1. Train-Test Split Shapes ---")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}\n")
```

```python
# ==========================================
# Q2. (Bag of Words – Basic)
# ==========================================
# Apply CountVectorizer for unigrams
bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)

print("--- Q2. Bag of Words (Basic) ---")
print(f"X_train_bow shape: {X_train_bow.shape}")
print(f"X_test_bow shape: {X_test_bow.shape}")
print("First 20 feature names:")
print(bow_vectorizer.get_feature_names_out()[:20])
print()
```

```python
# ==========================================
# Q3. (Bag of Words + MultinomialNB)
# ==========================================
# Train Multinomial Naive Bayes model using BoW features
nb_bow = MultinomialNB()
nb_bow.fit(X_train_bow, y_train)

# Make predictions and calculate accuracy
y_pred_bow = nb_bow.predict(X_test_bow)
accuracy_bow = accuracy_score(y_test, y_pred_bow)

print("--- Q3. BoW + MultinomialNB ---")
print(f"Bag of Words (Unigram) Test Accuracy: {accuracy_bow:.4f}\n")
```

```python
# ==========================================
# Q4. (Understanding Vocabulary)
# ==========================================
feature_names = bow_vectorizer.get_feature_names_out()

print("--- Q4. Understanding Vocabulary ---")
print(f"Total vocabulary size: {len(feature_names)}")
print(f"Sample 15 words from vocabulary: {list(feature_names[100:115])}")

# Convert one sample document into its BoW vector
sample_idx = 0
sample_text = X_train.iloc[sample_idx]
sample_vector = X_train_bow[sample_idx].toarray()

print(f"\nSample Document: '{sample_text}'")
print(f"Corresponding BoW Vector (non-zero entries):")
for idx, val in enumerate(sample_vector[0]):
    if val > 0:
        print(f"Word: '{feature_names[idx]}' -> Count: {val}")
print()
```

```python
# ==========================================
# Q5. (N-grams with Bag of Words)
# ==========================================
# Create CountVectorizer with unigrams and bigrams
ngram_vectorizer = CountVectorizer(ngram_range=(1, 2))
X_train_ngram = ngram_vectorizer.fit_transform(X_train)
X_test_ngram = ngram_vectorizer.transform(X_test)

print("--- Q5. N-grams with Bag of Words ---")
print(f"X_train_ngram shape: {X_train_ngram.shape}")
print(f"X_test_ngram shape: {X_test_ngram.shape}")
print("Sample bigram/unigram features:")
print(ngram_vectorizer.get_feature_names_out()[5000:5015])
print()
```

```python
# ==========================================
# Q6. (N-grams + Model Training)
# ==========================================
# Train MultinomialNB with n-gram features
nb_ngram = MultinomialNB()
nb_ngram.fit(X_train_ngram, y_train)

y_pred_ngram = nb_ngram.predict(X_test_ngram)
accuracy_ngram = accuracy_score(y_test, y_pred_ngram)

print("--- Q6. N-grams + Model Training ---")
print(f"Bag of Words (Unigram + Bigram) Test Accuracy: {accuracy_ngram:.4f}")
print(f"Comparison: Unigram Accuracy = {accuracy_bow:.4f} vs Unigram+Bigram Accuracy = {accuracy_ngram:.4f}\n")
```

```python
# ==========================================
# Q7. (TF-IDF Vectorization)
# ==========================================
# Apply TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

print("--- Q7. TF-IDF Vectorization ---")
print(f"X_train_tfidf shape: {X_train_tfidf.shape}")
print(f"X_test_tfidf shape: {X_test_tfidf.shape}")
print("First 15 TF-IDF feature names:")
print(tfidf_vectorizer.get_feature_names_out()[:15])
print()
```

```python
# ==========================================
# Q8. (TF-IDF + MultinomialNB)
# ==========================================
# Train MultinomialNB with TF-IDF features
nb_tfidf = MultinomialNB()
nb_tfidf.fit(X_train_tfidf, y_train)

y_pred_tfidf = nb_tfidf.predict(X_test_tfidf)
accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("--- Q8. TF-IDF + MultinomialNB ---")
print(f"TF-IDF Test Accuracy: {accuracy_tfidf:.4f}\n")
```

```python
# ==========================================
# Q9. (Comparison of Vectorizers)
# ==========================================
comparison_data = {
    "Vectorization Approach": [
        "Bag of Words (Unigrams)",
        "Bag of Words (Unigrams + Bigrams)",
        "TF-IDF"
    ],
    "Test Accuracy": [
        accuracy_bow,
        accuracy_ngram,
        accuracy_tfidf
    ]
}

comparison_df = pd.DataFrame(comparison_data)
print("--- Q9. Comparison Table ---")
print(comparison_df.to_string(index=False))
print("\nObservation:")
print("TF-IDF or N-grams generally capture term importance or contextual word pairs better than basic BoW.")
print("TF-IDF penalizes frequent words that carry less emotional weight, often yielding robust performance for text classification tasks.\n")
```

```python
# ==========================================
# Q10. (Mini Project – Complete Vectorization Pipeline)
# ==========================================
print("--- Q10. Mini Project Pipeline Execution ---")
# Best vectorizer and model selection based on accuracy comparison
best_accuracy = max(accuracy_bow, accuracy_ngram, accuracy_tfidf)

if best_accuracy == accuracy_tfidf:
    best_vectorizer = tfidf_vectorizer
    best_model = nb_tfidf
    best_name = "TF-IDF + MultinomialNB"
elif best_accuracy == accuracy_ngram:
    best_vectorizer = ngram_vectorizer
    best_model = nb_ngram
    best_name = "Bag of Words (Bigrams) + MultinomialNB"
else:
    best_vectorizer = bow_vectorizer
    best_model = nb_bow
    best_name = "Bag of Words (Unigrams) + MultinomialNB"

print(f"Best Performing Approach: {best_name} with Accuracy: {best_accuracy:.4f}")

# Save the best vectorizer and best model using joblib
joblib.dump(best_vectorizer, 'best_vectorizer.pkl')
joblib.dump(best_model, 'best_model.pkl')
print("Best vectorizer saved as 'best_vectorizer.pkl'")
print("Best model saved as 'best_model.pkl'")
```
